```python
import pandas as pd

data = {
    "Name": ["Akash", "Priya", "Rahul", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 75, 95],
    "City": ["Hyderabad", "Chennai", "Delhi", "Mumbai"]
}

df = pd.DataFrame(data)

print(df)

```

```python
#View the data
print(df.head())       # First 5 rows
print(df.tail())       # Last 5 rows
print(df.shape)        # Number of rows and columns
print(df.columns)      # Column names
print(df.index)        # Row index
print(df.dtypes)       # Data types of columns
print(df.info())       # Dataset information
print(df.describe())   # Statistical summary
```

```python
#Select rows and columns
print(df["Name"])                 # Single column
print(df[["Name", "Marks"]])      # Multiple columns

print(df.iloc[0])                 # First row by position
print(df.iloc[0:2])               # First two rows
print(df.loc[0])                  # Row with index label 0
print(df.loc[0:2, ["Name", "Marks"]])
```

```python
#Filter data
print(df[df["Marks"] > 80])

print(df[df["Age"] >= 21])

print(df[df["City"] == "Hyderabad"])

print(df[(df["Marks"] > 80) & (df["Age"] >= 21)])

print(df[df["Marks"].between(80, 95)])
```

```python
#Add, update, and delete columns
# Add a column
df["Grade"] = ["B", "A", "C", "A"]

# Add a calculated column
df["Bonus_Marks"] = df["Marks"] + 5

# Update values
df.loc[df["Name"] == "Akash", "Marks"] = 88

# Delete a column
df.drop(columns=["Bonus_Marks"], inplace=True)

print(df)
```

```python
#Handle missing values
data = {
    "Name": ["Akash", "Priya", "Rahul", "Sneha"],
    "Marks": [85, None, 75, None],
    "Age": [20, 21, None, 22]
}

df = pd.DataFrame(data)

print(df.isnull())           # Find missing values
print(df.isnull().sum())     # Count missing values

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

```python
#Sort data
print(df.sort_values(by="Marks"))

print(df.sort_values(by="Marks", ascending=False))

print(df.sort_values(by="Age", ascending=True))
```

```python
#Calculate statistics
print(df["Marks"].mean())    # Average
print(df["Marks"].sum())     # Total
print(df["Marks"].max())     # Maximum
print(df["Marks"].min())     # Minimum
print(df["Marks"].median())  # Median
print(df["Marks"].count())   # Non-missing count
print(df["Marks"].std())     # Standard deviation
```

```python
#Group data
data = {
    "Name": ["A", "B", "C", "D", "E"],
    "Department": ["CSE", "AIML", "CSE", "AIML", "CSE"],
    "Marks": [80, 90, 70, 85, 95]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Marks"].mean())

print(df.groupby("Department")["Marks"].sum())

print(df.groupby("Department")["Marks"].count())
```

```python
#Remove duplicates
data = {
    "Name": ["Akash", "Priya", "Akash", "Rahul"],
    "Marks": [80, 90, 80, 75]
}

df = pd.DataFrame(data)

print(df.duplicated())

df = df.drop_duplicates()

print(df)
```

```python
#Apply functions and work with strings
df["Marks_Doubled"] = df["Marks"].apply(lambda x: x * 2)

df["Name_Upper"] = df["Name"].str.upper()

df["Name_Lower"] = df["Name"].str.lower()

print(df)
```

```python
#Combine DataFrames
df1 = pd.DataFrame({
    "Name": ["Akash", "Priya"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Rahul", "Sneha"],
    "Marks": [75, 95]
})

# Combine rows
result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

```python
#Merge DataFrames
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Akash", "Priya", "Rahul"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 75]
})

result = pd.merge(df1, df2, on="ID")

print(result)
```
