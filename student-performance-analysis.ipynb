```python
import pandas as pd
import numpy as np
```

```python
#Create the dataset
data = {
    "Name": ["Akash", "Priya", "Rahul", "Sneha", "Arjun",
             "Kavya", "Rohan", "Anjali", "Vijay", "Meena"],

    "Maths": [85, 92, 67, 95, 74, 88, 56, 90, 72, 96],
    "Science": [78, 89, 70, 94, 80, 91, 60, 87, 75, 98],
    "English": [82, 95, 65, 90, 78, 85, 62, 92, 70, 94],
    "Attendance": [90, 95, 75, 98, 80, 92, 65, 96, 78, 99]
}

df = pd.DataFrame(data)

df
```

```python
#Explore the data
print("First five rows:")
display(df.head())

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe())
```

```python
#Calculate total and average marks
df["Total"] = df[["Maths", "Science", "English"]].sum(axis=1)

df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

display(df)
```

```python
#Assign grades
def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(calculate_grade)

display(df[["Name", "Average", "Grade"]])
```

```python
#Find the top-performing students
top_students = df.sort_values(
    by="Average",
    ascending=False
)

display(top_students[["Name", "Average", "Grade"]].head(3))
```

```python
#Find students who need improvement
low_performers = df[df["Average"] < 70]

print("Students who need improvement:")
display(low_performers[["Name", "Average", "Grade"]])
```

```python
#Compare subject averages
subjects = ["Maths", "Science", "English"]

subject_averages = df[subjects].mean()

print("Average marks in each subject:")
display(subject_averages)
```

```python
#Analyze attendance
low_attendance = df[df["Attendance"] < 75]

print("Students with attendance below 75%:")
display(low_attendance[["Name", "Attendance"]])
```

```python
#Export the analyzed dataset
df.to_csv("student_performance_results.csv", index=False)

print("Analysis saved successfully!")
```
